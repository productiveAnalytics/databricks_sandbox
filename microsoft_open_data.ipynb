```
# import
from pyspark.sql import SparkSession
```

```
# Spark session
spark = SparkSession.builder.getOrCreate()
```

```
# connect to Azure blob storage
container_name = 'tutorials'
account_name = 'azureopendatastorage'
relative_path = 'noaa_isd_weather/demo.parquet'
wasbs_path = f'wasbs://{container_name}@{account_name}.blob.core.windows.net/{relative_path}'
```

```
df = spark.read.parquet(wasbs_path)
df.count()
```

```
print('Register the Dataframe as SQL temp table')
df.createOrReplaceTempView('weather_data')
```

```
%sql
select datetime, wd.`stations.city`, wd.count as record_count from weather_data wd
```
